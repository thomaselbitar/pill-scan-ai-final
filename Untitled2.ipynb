```python
import os
import shutil
import random
from pathlib import Path

RAW_DATA_DIR = Path("/content/drive/MyDrive/PharmacyProject/dataset_raw/dataset")
OUT_DIR = Path("/content/drive/MyDrive/PharmacyProject/dataset_split")

TRAIN_RATIO = 0.7
VAL_RATIO = 0.15   # test = 0.15
random.seed(42)

OUT_DIR.mkdir(parents=True, exist_ok=True)

splits = ["train", "val", "test"]

for split in splits:
    (OUT_DIR / split).mkdir(exist_ok=True)

for class_dir in RAW_DATA_DIR.iterdir():
    if not class_dir.is_dir():
        continue

    class_name = class_dir.name
    print("Processing:", class_name)

    images = [p for p in class_dir.iterdir() if p.is_file()]
    random.shuffle(images)

    n = len(images)
    n_train = int(TRAIN_RATIO * n)
    n_val = int(VAL_RATIO * n)
    n_test = n - n_train - n_val

    train_imgs = images[:n_train]
    val_imgs   = images[n_train:n_train+n_val]
    test_imgs  = images[n_train+n_val:]

    for split in splits:
        (OUT_DIR / split / class_name).mkdir(parents=True, exist_ok=True)

    def copy_files(files, split):
        for img in files:
            dst = OUT_DIR / split / class_name / img.name
            if not dst.exists():
                shutil.copy2(img, dst)

    copy_files(train_imgs, "train")
    copy_files(val_imgs, "val")
    copy_files(test_imgs, "test")

print("Done! Dataset is split into Train/Val/Test.")

# Show the result
!ls -R "/content/drive/MyDrive/PharmacyProject/dataset_split"
```

```python
from pathlib import Path
import torch
from torch import nn, optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

DATA_ROOT = Path("/content/drive/MyDrive/PharmacyProject/dataset_split")
TRAIN_DIR = DATA_ROOT / "train"
VAL_DIR = DATA_ROOT / "val"
TEST_DIR = DATA_ROOT / "test"

print("Train folder:", TRAIN_DIR)
print("Val folder:", VAL_DIR)
print("Test folder:", TEST_DIR)
```

```python
# ImageNet standard transforms
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

# Datasets
train_dataset = datasets.ImageFolder(root=str(TRAIN_DIR), transform=train_transform)
val_dataset   = datasets.ImageFolder(root=str(VAL_DIR), transform=val_test_transform)
test_dataset  = datasets.ImageFolder(root=str(TEST_DIR), transform=val_test_transform)

# DataLoaders
BATCH_SIZE = 32

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

class_names = train_dataset.classes
num_classes = len(class_names)

print("Number of classes:", num_classes)
print("Example classes:", class_names[:10])
```

```python
from torchvision.models import mobilenet_v3_large, MobileNet_V3_Large_Weights

# Load pretrained ImageNet weights
weights = MobileNet_V3_Large_Weights.DEFAULT
model = mobilenet_v3_large(weights=weights)

# Replace classifier last layer with correct number of classes
in_features = model.classifier[-1].in_features
model.classifier[-1] = nn.Linear(in_features, num_classes)

model = model.to(device)

# Define loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

print("Model ready.")
print("Number of classes:", num_classes)
```

```python
import time
import copy
import torch

def train_model(model, criterion, optimizer, train_loader, val_loader, num_epochs=10):
    start_time = time.time()
    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch+1}/{num_epochs}")
        print("-" * 20)

        for phase in ["train", "val"]:
            if phase == "train":
                model.train()
                loader = train_loader
            else:
                model.eval()
                loader = val_loader

            running_loss = 0.0
            running_corrects = 0
            total = 0

            for inputs, labels in loader:
                inputs = inputs.to(device)
                labels = labels.to(device)

                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == "train"):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    if phase == "train":
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
                total += labels.size(0)

            epoch_loss = running_loss / total
            epoch_acc = running_corrects.double() / total

            print(f"{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

            # Save best model
            if phase == "val" and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())

    total_time = time.time() - start_time
    print(f"\nTraining Complete in {total_time/60:.1f} minutes")
    print(f"Best Val Accuracy: {best_acc:.4f}")

    model.load_state_dict(best_model_wts)
    return model
```

```python
import torch
import json
from pathlib import Path

SAVE_DIR = Path("/content/drive/MyDrive/PharmacyProject/model_v1")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# Save model weights
model_path = SAVE_DIR / "mobilenetv3_pharmacy.pth"
torch.save(model.state_dict(), model_path)
print("Model saved to:", model_path)

# Save class names
labels_path = SAVE_DIR / "class_names.json"
with open(labels_path, "w") as f:
    json.dump(class_names, f)
print("Class names saved to:", labels_path)
```

```python
# Load class names
with open(labels_path, "r") as f:
    loaded_classes = json.load(f)

# Recreate model
loaded_model = mobilenet_v3_large(weights=None)
in_features = loaded_model.classifier[-1].in_features
loaded_model.classifier[-1] = nn.Linear(in_features, len(loaded_classes))
loaded_model.load_state_dict(torch.load(model_path, map_location=device))
loaded_model = loaded_model.to(device)
loaded_model.eval()

print("Model reloaded successfully!")
```

```python
from PIL import Image

def predict_image(model, img_path, class_names):
    img = Image.open(img_path).convert("RGB")
    img_t = val_test_transform(img).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = model(img_t)
        probs = torch.softmax(outputs, dim=1)
        conf, pred_idx = torch.max(probs, 1)

    return class_names[pred_idx.item()], float(conf)

pred, confidence = predict_image(loaded_model, test_image_path, loaded_classes)
print("Prediction:", pred)
print("Confidence:", confidence)
```

```python
import torch
from torch.utils.data import DataLoader

def evaluate_test_accuracy(model, test_loader):
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in test_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            _, preds = torch.max(outputs.data, 1)

            total += labels.size(0)
            correct += (preds == labels).sum().item()

    accuracy = correct / total
    print(f"Test Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
    return accuracy

# Run evaluation
test_accuracy = evaluate_test_accuracy(model, test_loader)
```
